## Mission Overview
**Quantum Heist: Break the Entanglement Vault** is a strategic, mission-based quantum puzzle game where players act as an elite quantum infiltrator tasked with breaching three security layers of the high-tech Entanglement Vault:
1. **Door 1 (Laser Grid):** Reach deterministic target state $|1\rangle$ using single-qubit gates ($X$) within a 2-move limit.
2. **Door 2 (Dual-Sensor):** Create balanced superposition $\frac{|0\rangle + |1\rangle}{\sqrt{2}}$ using Hadamard ($H$) within a 2-move limit, avoiding self-cancellation ($H \cdot H = I$).
3. **Door 3 (Master Vault):** Synchronize two spatially separated locks into Bell State $|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$ using $H$ and $CNOT$ with 0% mismatch within a 3-move limit.

All puzzle outcomes are determined by genuine Qiskit circuit execution and measurement simulation.

## 1. Setup & Environment Verification

```python
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
try:
    from qiskit_aer import AerSimulator
    simulator = AerSimulator()
    backend_name = "AerSimulator"
except ImportError:
    from qiskit.primitives import StatevectorSampler
    simulator = StatevectorSampler()
    backend_name = "StatevectorSampler"

print("Qiskit version verified!")
print(f"Using backend: {backend_name}")
```

## 2. Door 1: Security Laser Grid (Target State $|1\rangle$)
- **Objective:** Transform qubit from initial ground state $|0\rangle$ into target state $|1\rangle$.
- **Move Limit:** Max 2 moves.
- **Solution:** Apply Pauli-$X$ gate ($X|0\rangle = |1\rangle$).

```python
qc1 = QuantumCircuit(1, 1)
qc1.x(0)  # Pauli-X (Quantum NOT)
qc1.measure(0, 0)

print("Door 1 Circuit Diagram:")
print(qc1.draw(output="text"))

counts1 = simulator.run(qc1, shots=200).result().get_counts()
print(f"\nDoor 1 Measurement Counts (200 shots): {counts1}")
if counts1.get('1', 0) == 200:
    print("Outcome: 100% of shots collapsed to target state |1>. Door 1 Bypassed!")

# Plot measurement distribution
fig, ax = plt.subplots(figsize=(5, 3), facecolor='#0F172A')
ax.set_facecolor('#0F172A')
ax.bar(list(counts1.keys()), [v/200*100 for v in counts1.values()], color='#38BDF8', width=0.4)
ax.set_ylabel('Probability (%)', color='#E2E8F0')
ax.set_title('Door 1: Target State |1> Measurement Counts', color='#F8FAFC', fontweight='bold')
ax.tick_params(colors='#E2E8F0')
ax.set_ylim(0, 110)
plt.show()
```

## 3. Door 2: The Quantum Dual-Sensor (Superposition)
- **Objective:** Put qubit into balanced superposition $|\psi\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}}$.
- **Move Limit:** Max 2 moves.
- **Solution:** Apply Hadamard ($H$) gate.

```python
qc2 = QuantumCircuit(1, 1)
qc2.h(0)  # Hadamard creates superposition
qc2.measure(0, 0)

print("Door 2 Superposition Circuit:")
print(qc2.draw(output="text"))

counts2 = simulator.run(qc2, shots=200).result().get_counts()
print(f"\nDoor 2 Measurement Counts (200 shots): {counts2}")
p0 = (counts2.get('0', 0) / 200) * 100
p1 = (counts2.get('1', 0) / 200) * 100
if p0 >= 35.0 and p1 >= 35.0:
    print(f"Outcome: Balanced superposition observed (|0>: {p0:.1f}%, |1>: {p1:.1f}%). Door 2 Deactivated!")

# Plot measurement distribution
fig, ax = plt.subplots(figsize=(5.5, 3), facecolor='#0F172A')
ax.set_facecolor('#0F172A')
ax.bar(list(counts2.keys()), [v/200*100 for v in counts2.values()], color=['#38BDF8', '#818CF8'], width=0.45)
ax.set_ylabel('Probability (%)', color='#E2E8F0')
ax.set_title('Door 2: Superposition (|0> + |1>)/sqrt(2) Distribution', color='#F8FAFC', fontweight='bold')
ax.tick_params(colors='#E2E8F0')
ax.set_ylim(0, 75)
plt.show()
```

### Self-Interference Trap ($H \cdot H = I$)
If an infiltrator over-applies $H$ twice, destructive interference cancels out the superposition and returns the qubit to ground state $|0\rangle$:

```python
qc2_trap = QuantumCircuit(1, 1)
qc2_trap.h(0)
qc2_trap.h(0)
qc2_trap.measure(0, 0)

counts_trap = simulator.run(qc2_trap, shots=200).result().get_counts()
print(f"Double Hadamard Counts: {counts_trap}")
print("Physics: H * H = Identity! State returned to |0>. Superposition failed.")
```

## 4. Door 3: The Entangled Master Vault (Bell State $|\Phi^+\rangle$)
- **Objective:** Create Bell State $|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$ across Lock A ($q_0$) and Lock B ($q_1$).
- **Move Limit:** Max 3 operations.
- **Constraint:** 0% mismatch tolerance ('01' or '10').
- **Solution:** Apply $H(q_0)$ followed by $CNOT(q_0 \to q_1)$.

```python
qc3 = QuantumCircuit(2, 2)
qc3.h(0)      # Superposition on Lock A
qc3.cx(0, 1)  # CNOT entangles Lock A and Lock B
qc3.measure(0, 0)
qc3.measure(1, 1)

print("Door 3 Bell State Circuit:")
print(qc3.draw(output="text"))

counts3 = simulator.run(qc3, shots=200).result().get_counts()
print(f"\nDoor 3 Measurement Counts (200 shots): {counts3}")
mismatches = counts3.get('01', 0) + counts3.get('10', 0)
print(f"Mismatches ('01' or '10'): {(mismatches / 200) * 100:.1f}%")
if mismatches == 0 and counts3.get('00', 0) > 0 and counts3.get('11', 0) > 0:
    print("Outcome: 100% correlation achieved! Master Vault Cracked!")

# Plot measurement distribution
fig, ax = plt.subplots(figsize=(6, 3), facecolor='#0F172A')
ax.set_facecolor('#0F172A')
states = ['00', '01', '10', '11']
counts_all = [counts3.get(s, 0) for s in states]
ax.bar(states, [c/200*100 for c in counts_all], color=['#10B981', '#EF4444', '#EF4444', '#10B981'], width=0.5)
ax.set_ylabel('Probability (%)', color='#E2E8F0')
ax.set_title('Door 3: Bell State |Phi+> Correlation (0% Mismatch)', color='#F8FAFC', fontweight='bold')
ax.tick_params(colors='#E2E8F0')
ax.set_ylim(0, 75)
plt.show()
```
